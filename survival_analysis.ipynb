# Part B — risk stratification, forest plots and covariate curves
Runs after `clean.py` and `baseline.ipynb`. Uses the same preprocessing, 10-fold CV seed and settings (100 genes, penaliser 1.0). Figures are saved to `reports/figures/`.

```
%matplotlib inline
import json
import survival_common as sc
import survival_analysis as sa
sets = sc.feature_sets()
print({k: v.shape for k, v in sets.items()})
```

## 1. Kaplan–Meier: high vs low risk + log-rank test
Risk = **out-of-fold** log-hazard (each patient scored by a model that never saw them), split at the median. Using in-sample scores here would exaggerate the separation.

```
fig, res = sa.km_high_low()
print(json.dumps(res, indent=2))
```

## 2. Forest plots (hazard ratios, 95% CI)
Clinical model in natural units; combined model per +1 SD, showing the 15 strongest features. CIs come from the penalised fit, so treat them as approximate.

```
f1, f2, t_clin, t_all = sa.forest_plots(sets)
t_clin.round(3)
```

## 3. Covariate survival curves
Unadjusted KM by stage, then model-based curves varying one feature at a time.

```
fig, p_stage = sa.km_by_stage()
print('stage log-rank p =', round(p_stage, 4))
```

```
fig, top_genes = sa.covariate_curves(sets)
top_genes
```
